# Feature Engineering

```python
df.info()
```

```python
df_encoded = df.copy()
```

```python
df['department'].value_counts()
```

```python
# Frequency Encoding
department_count = df_encoded['department'].value_counts().to_dict()
df_encoded['department'] = df_encoded['department'].map(department_count)
```

```python
df['region'].value_counts()
```

```python
region_count = df_encoded['region'].value_counts().to_dict()
df_encoded['region'] = df_encoded['region'].map(region_count)
```

```python
df['education'].value_counts()
```

```python
# Label Encoding
df_encoded['education'] = df_encoded['education'].map({'High School':0, 'Below Secondary':1,"Bachelor's":2,"Master's":3,"Master's & above":4})
```

```python
# One-hot using dummy
df_encoded = pd.get_dummies(df_encoded, columns=['gender'], drop_first=True, dtype=int)
```

```python
df_encoded['recruitment_channel'].value_counts()
```

```python
df_encoded = pd.get_dummies(df_encoded, columns=['recruitment_channel'], drop_first=True, dtype=int)
```
