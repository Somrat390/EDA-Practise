```python
dfSE = pd.concat([df1SE, df2SE], ignore_index=True)
dfML = pd.concat([df1ML, df2ML], ignore_index=True)
dfML
```

# Student ID has two column and repetitive also we have to fix this one  by Using df.merge with an inner join

```python
dfSE = pd.concat([df1SE, df2SE], ignore_index=True)
dfML = pd.concat([df1ML, df2ML], ignore_index=True)

df = dfSE.merge(dfML, how='inner')
df
```

# So in merge during inner jon it only take the student who are both present in both courses

So there is 21 student who participate in both course

# Join
* The inner join takes the intersection from two or more dataframes. It corresponds to the INNER JOIN in Structured Query Language (SQL).
* The outer join takes the union from two or more dataframes. It corresponds to the FULL OUTER JOIN in SQL.
* The left join uses the keys from the left-hand dataframe only. It corresponds to the LEFT OUTER JOIN in SQL.
* The right join uses the keys from the right-hand dataframe only. It corresponds to the RIGHT OUTER JOIN in SQL

```python
df = dfSE.merge(dfML, how='left')
df
```

So from this we can tell only 26 students praticipate in Software Engineering course

```python
df = dfSE.merge(dfML, how='right')
df
```

So 25 student take participate in Machine Learning course

```python
df = dfSE.merge(dfML, how='outer')
df
```

# By outer join we can identify 30 student praticipate atleast one course

# Merging on index

```python
left1 = pd.DataFrame({'key': ['apple','ball','apple', 'apple', 'ball', 'cat'], 'value': range(6)})
right1 = pd.DataFrame({'group_val': [33.4, 5]}, index=['apple', 'ball'])
```

```python
df = pd.merge(left1, right1, left_on='key', right_index=True, how='outer')
df
```

```python
df = pd.merge(left1, right1, left_on='key', right_index=True)
df
```

# Reshaping and pivoting
Stacking: Stack rotates from any particular column in the data to the rows.
Unstacking: Unstack rotates from the rows into the column

```python
data = np.arange(15).reshape((3,5))
indexers = ['Rainfall', 'Humidity', 'Wind']
dframe1 = pd.DataFrame(data, index=indexers, columns=['Bergen', 'Oslo', 'Trondheim', 'Stavanger', 'Kristiansand'])
dframe1
```

 using the stack() method on the preceding dframe1, we can pivot the columns into rows to produce a series:

```python
stacked = dframe1.stack()
stacked
```

```python
stacked.unstack()
```

```python
series1 = pd.Series([000, 111, 222, 333], index=['zeros','ones', 'twos', 'threes'])
series2 = pd.Series([444, 555, 666], index=['fours', 'fives', 'sixes'])

frame2 = pd.concat([series1, series2], keys=['Number1', 'Number2'])
frame2
```
